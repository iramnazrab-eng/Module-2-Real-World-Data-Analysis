```python
df = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin1")

df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())
```

```python
df["Category"] = df["Category"].fillna("Unknown")
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Order Date"].head()
```

```python
df["Month"] = df["Order Date"].dt.month

df["Year"] = df["Order Date"].dt.year

df[["Order Date", "Month", "Year"]].head()
```

```python
df["Order Month"] = df["Order Date"].dt.to_period("M")

df[["Order Date", "Order Month"]].head()
```

```python
product_sales = df.groupby("Product Name")["Sales"].sum()

top_products = product_sales.sort_values(
    ascending=False
).head(10)

print(top_products)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales.sort_values(ascending=False))
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

print(category_profit.sort_values(ascending=False))
```

```python
monthly_sales = df.groupby("Order Month")["Sales"].sum()

print(monthly_sales.sort_values(ascending=False).head(10))
```

```python
customer_orders = df.groupby("Customer ID")["Order ID"].nunique()

repeat_customers = customer_orders[customer_orders > 1]

print("Total customers:", customer_orders.count())

print("Repeat customers:", repeat_customers.count())
```

```python
repeat_percentage = (
    repeat_customers.count() / customer_orders.count()
) * 100

print("Repeat customer percentage:", round(repeat_percentage, 2), "%")
```

```python
correlation = df[["Sales", "Profit", "Quantity"]].corr()

print(correlation)
```

```python
plt.figure(figsize=(6, 4))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Between Sales, Profit and Quantity")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```
